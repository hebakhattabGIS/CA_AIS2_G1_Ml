```python
email = input("Please enter your email address: ")
```

#run a few checks on the user input

```python
print(f"Validating email: {email}")


# Check if email has exactly one "@"
at_count = email.count("@")

# Check if email has "." in the last 4 letters
has_dot_in_end = "." in email[-4:]

# Validate email
if at_count == 1 and has_dot_in_end:
    print("Email is valid!")
else:
    print("Email is invalid!")
    if at_count != 1:
        print(f"  - Must have exactly one '@', found {at_count}")
    if not has_dot_in_end:
        print("  - Must have '.' in the last 3 characters")
```

```python
#extract username from email 
username = email.split("@")[0]
print(f"Username: {username}")
#extract domain from email
domain = email.split("@")[1]
print(f"Domain: {domain}")
```

```python
if domain.endswith(".com"):
    domain_category = "Commercial Domain"
elif domain.endswith(".edu"):
    domain_category = "Educational Domain"
else:
    domain_category = "Other Domain"

print(domain_category)
```
