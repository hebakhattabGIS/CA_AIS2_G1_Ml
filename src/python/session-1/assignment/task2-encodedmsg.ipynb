```python
encodedmsg="###!!@mocleW EPGTQ!!!6789"
```

```python
words = ''.join(ch for ch in encodedmsg if ch.isalpha() or ch.isspace())
print(words)
```

```python
first, rest = words.split(' ', 1)
reversed_first = first[::-1]
s = reversed_first + ' ' + rest
i = len(reversed_first)
new_words = s[:i] + s[i+1] + s[i] + s[i+2:]
first_word, second_word = new_words.split(' ', 1)
new_words = first_word.capitalize() + ' ' + second_word
print(new_words)
```
