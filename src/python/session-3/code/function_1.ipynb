```python
def info_user(name : str = "heba", age : int | None = 10) -> str:
# type hist, and default values are specified. age paramter here is optional and the default value is 10
    '''
    This function takes a name and age as input and returns a formatted string.
    Parameters:
    name (str): The name of the user.
    age (int): The age of the user.
    Returns:
    str: A formatted string containing the name and age.
    '''
    return f"my name is {name} and I am {age} years old."
```

```python
def func(item,lst=None):
    if lst is None:
        lst = []
    lst.append(item)
    return lst

print(func("1"))
print(func("2"))
print(func("3"))
#don't use mutable objects like list, dict, set as default arguments. use none and then create the object inside the function when needed.
```

```python
def train_model(model_name, *datasets, **hyperparameters):
    print(f"Training model: {model_name}")

    print("\nDatasets:") 
    for dataset in datasets:
            print("-", dataset)
    print("\nhyperparameters:")
    for key, value in hyperparameters.items():
        print(f"{key}: {value}")
```

```python
train_model(
    "fraud detection",
    "dataset1.csv",
    "validation.csv",
    "test.csv",
    learning_rate=0.001,
    epochs=100,
    batch_size=64,
    optimizer="Adam",
    early_stopping=True
)
```

```python
transformers = {
    "lower":lambda text:text.lower(),
    "upper":lambda text:text.upper(),
    "strip":lambda text:text.strip(),
    "title":lambda text:text.title(),
    "reverse":lambda text:text[::-1],
    "word_count":lambda text:len(text.split()),
    "char_count":lambda text:len(text),
    "remove_spaces":lambda text:text.replace(" ",""),
    
}
```

```python
text = "   Machine Learning With Python "

print(transformers["strip"](text))
print(transformers["lower"](text))
print (transformers["title"](text))
```
