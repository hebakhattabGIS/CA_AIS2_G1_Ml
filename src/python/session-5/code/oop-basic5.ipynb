```python
class Student:

    def __init__(self, name, age) -> None:
       self._name = name  #protected attribute
       self._age = age

    @property
    def name(self):
       return self._name

    @property       #getter must be defined 
    def age(self):
        return self._age
    @age.setter     #setter 
    def age(self, value):
        if value < 0 :
            raise ValueError("age must be positive")
        self._age = value

 
```

```python
s1 = Student("heba", 20)
s1.age = 10
s1.age
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Bird(Animal):

    def move(self):
        print("move from bird")

class Cat(Animal):

    def move(self):
        print("move from cat")


a = Bird()
a.move()
```

```python
#polymorphosim
class Shape:
    def calculate_area(self):
        pass

class Circle(Shape):
    def __init__(self, radius) -> None:
        self.radius = radius

    def calculate_area(self):
        return 3.14 * self.radius *self.radius

class Square(Shape):
    def __init__(self, side) -> None:
        self.side = side

    def calculate_area(self):
        return self.side **2

if __name__ == "__main__":
    x = Circle(5)
    y = Square(4)
    print("Area circle:", x.calculate_area())
    print("Area Square:", y.calculate_area())
```
