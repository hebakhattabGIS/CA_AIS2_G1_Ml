```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, root_mean_squared_error
import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv('Salary_Data.csv')
```

```python
data.head()
```

```python
plt.Figure(figsize=(3,3))
sns.pairplot(data, x_vars=['YearsExperience'], y_vars=['Salary'], size=3, kind="scatter")
plt.xlabel("years")
plt.ylabel("salary")
plt.title("salary prediction")
plt.show()
```

```python
x = data.iloc[:,:-1]
y = data.iloc[:,1]

x_train, x_test, y_train, y_test = train_test_split(x,y, train_size=0.8,random_state=50)
```

```python
y_train
```

```python
lin_reg = LinearRegression()
lin_reg.fit(x_train, y_train)
y_hat_train = lin_reg.predict(x_train)
```

```python
plt.scatter(x_train, y_train, color= 'red')
plt.plot(x_train, y_hat_train, color = 'blue')
plt.title("Salary vs experience of years")
plt.xlabel("years")
plt.ylabel("Salary")
plt.show()
```

```python
y_hat_test = lin_reg.predict(x_test)
```

```python
plt.scatter(x_test, y_test, color= 'red')
plt.plot(x_test, y_hat_test, color = 'blue')
plt.title("Salary vs experience of years")
plt.xlabel("years")
plt.ylabel("Salary")
plt.show()
```

```python
#calculation of mse
mse = mean_squared_error(y_test, y_hat_test)
print("mse is ", mse)
#root mse
rmse = root_mean_squared_error(y_test, y_hat_test)
print(rmse)
#r2 score
r2 = r2_score(y_train, y_hat_train)
r2
```

```python
w = lin_reg.coef_
b = lin_reg.intercept_
#print(w, b)
y_pred = w * x + b
input_x = int(input("enter your years of experience"))
y_pred = w * input_x +b
print(y_pred)
```
