```python
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
x = np.array([[1],[2],[3],[4]])
y = np.array([2,4,6,8])
#y = 2x  w=2  b = 0
#linear Regression---> overfitting
lin_reg = LinearRegression()
lin_reg.fit(x,y)
w_lin = lin_reg.coef_
b_lin = lin_reg.intercept_
y_pred_lin = lin_reg.predict(x)

#ridge regression--->used to fix overfitting
ridge = Ridge(alpha=1.0)
ridge.fit(x,y)
w_ridge = ridge.coef_
b_ridge = ridge.intercept_
y_pred_ridge = ridge.predict(x)

#lasso regression--->used to fix overfitting
lasso = Lasso(alpha=1.0)
lasso.fit(x,y)
w_lasso = lasso.coef_
b_lasso = lasso.intercept_
y_pred_lasso = lasso.predict(x)

print("linear Regression")
print("mse : ", mean_squared_error(y,y_pred_lin))
print("R2 : ", r2_score(y,y_pred_lin))
print("W ", w_lin)
print("b ", b_lin)
print("="*40)
print("Ridge Regression")
print("mse : ", mean_squared_error(y,y_pred_ridge))
print("R2 : ", r2_score(y,y_pred_ridge))
print("W ", w_ridge)
print("b ", b_ridge)
print("="*40)
print("Lasso Regression")
print("mse : ", mean_squared_error(y,y_pred_lasso))
print("R2 : ", r2_score(y,y_pred_lasso))
print("W ", w_lasso)
print("b ", b_lasso)
print("="*40)
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.scatter(x,y, label="Actual data")
plt.plot(x,y_pred_lin, color = 'red',label ="Linear prediction")
plt.title("LinearRegression")
plt.legend()

plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
#plt.scatter(x,y, label="Actual data")
plt.plot(x,y_pred_lin, color = 'red',label ="Linear prediction")
plt.plot(x,y_pred_ridge, color = 'green',label ="ridge prediction")
plt.plot(x,y_pred_lasso, color = 'purple',label ="lasso prediction")
plt.title("LinearRegression")
plt.legend()
 
```
